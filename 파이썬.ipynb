## ✅ 문제 3. 추천 데이터 출력 함수 만들기

- MBTI별 추천 정보를 출력해 봅시다.
- `print_recommendation(user_info)` 함수를 만듭니다.
    - 함수명과 변수명을 적당하게 바꿔도 좋습니다.
- 해당 MBTI가 `recommendations`에 있다면 추천 항목을 출력하고, 없다면 안내 문구를 출력합니다.

> 출력 결과 예시:
- 해당 MBTI가 있다면,
```
[취미] 글쓰기, 산책  
[음악] 발라드
```
- 해당 MBTI가 없다면,
```
해당 MBTI에 대한 추천 정보가 없습니다.
```

````python
# 여기에 코드를 작성하세요
def print_recommendation(user_info):
    name, mbti = user_info
    if mbti in recommendations:
        for category, items in recommendations[mbti]:
            print(f"[{category}] {', '.join(items)}")
    else:
        print("해당 MBTI에 대한 추천 정보가 없습니다.")

print_recommendation(("서원민", "INTP"))
````

## ✅ 문제 4. 추천 항목 추가 함수 만들기

- `add_recommendation(mbti, category, item)` 함수를 만듭니다.
    - 함수명과 변수명을 적당하게 바꿔도 좋습니다.
- 해당 MBTI가 없으면 새로 만들고, 카테고리가 없다면 새로 추가합니다.

````python
# 여기에 코드를 작성하세요
def add_recommendation(mbti, category, item):
    mbti = mbti.upper()
    if mbti not in recommendations:
        recommendations[mbti] = []
    for cat, items in recommendations[mbti]:
        if cat == category:
            items.append(item)
            return
    recommendations[mbti].append((category, [item]))

add_recommendation("INTP", "음악", "재즈")
print_recommendation(("서원민", "INTP"))
````

<details>

<summary>💡 힌트 보기 (눌러서 열기)</summary>

- `recommendations` 딕셔너리에 MBTI가 **없는 경우**, 새로운 키로 추가해야 합니다.
- MBTI가 있다면, 그 안의 리스트에서 **해당 카테고리를 찾아야** 합니다.
- 이미 존재하는 카테고리라면, 항목을 그 리스트에 **덧붙이는 방식(append)**으로 추가합니다.
- 카테고리가 없다면, `(카테고리, [항목])` 형태의 튜플을 리스트에 추가해보는 건 어떨까요?

## ✅ 문제 6. 콘솔 앱처럼 메뉴 구성하기

- 아래와 같은 메뉴를 출력하고, 사용자로부터 반복적으로 입력을 받아 기능을 수행하는 프로그램을 만들어봅시다.
- `while True` 반복문과 `if` 조건문을 활용합니다.
- 각 기능은 함수로 따로 만들어 호출하는 것이 좋습니다. 위에서 만든 함수를 활용해볼까요?
- 사용자가 4번을 선택하면 프로그램이 종료되어야 합니다.

> 출력 결과 예시:
```
[ MBTI 취향 찾기 메뉴 ]
1. 추천 보기
2. 추천 추가
3. 사용자 목록 보기
4. 종료
```
```
번호를 입력하세요: 1
```

````python

# 여기에 코드를 작성하세요
def show_recommendation():
    name = input("이름을 입력하세요: ")
    mbti = input("MBTI를 입력하세요: ").upper()
    user = (name, mbti)
    users.append(user)
    print_recommendation(user)           # 문제 3 함수

def add_menu():
    mbti = input("MBTI: ")
    category = input("카테고리: ")
    item = input("추천 항목: ")
    add_recommendation(mbti, category, item)   # 문제 4 함수
    print("추가되었습니다.")

def show_users():
    for user in users:                   # 문제 5 리스트
        print(f"\n{user[0]}님 ({user[1]})")
        print_recommendation(user)

while True:
    print("\n[ MBTI 취향 찾기 메뉴 ]")
    print("1. 추천 보기")
    print("2. 추천 추가")
    print("3. 사용자 목록 보기")
    print("4. 종료")
    choice = input("번호를 선택하세요: ")

    if choice == "1":
        show_recommendation()
    elif choice == "2":
        add_menu()
    elif choice == "3":
        show_users()
    elif choice == "4":
        print("프로그램을 종료합니다.")
        break
    else:
        print("1~4 중에서 선택하세요.")
````

<details>

<summary> 💡 힌트 (눌러서 열기): </summary>

- 무한 반복을 위해 `while True:`를 사용할 수 있습니다.
- `input()`으로 사용자로부터 메뉴 번호를 입력받고, 조건문으로 분기합니다.
- 메뉴 번호가 1이면 추천 보기, 2면 추천 추가, 3이면 사용자 목록 보기, 4면 종료입니다.
- 종료하려면 `break`를 사용할 수 있습니다.
- `if`, `elif`, `else`를 활용해서 메뉴를 분기합니다.
- 추천 보기는 이전에 만든 `print_recommendation()` 함수를 이용할 수 있어요!

</details>
